# Etap 3, część 2 — bilans masy doby

**R = [W(koniec) − W(początek)] − Σ zasypów + Σ odsypów** dla każdej doby. Gdyby wszystkie masy były dokładne i wskazanie nie dryfowało, R = 0. R zbiera dryf wskazań i błędy mas zdarzeń w ciągu doby.

- **noc–noc** (główny): W(początek) = mediana wskazań 0:00–4:00 doby D, W(koniec) = mediana 0:00–4:00 doby D + 1.
- **5–22** (dodatkowy): W(początek) = mediana 4:50–5:00, W(koniec) = mediana 22:00–22:10 tej samej doby.
- Doba ważna, gdy oba okna i cały przedział leżą w jednej części okresu (bez luk > 60 s), a w oknach nie ma zdarzeń.
- Odsypy: wszystkie prawdziwe (przyjęte przez reguły i przyjęte w przeglądzie), masa z mediany 3 s, d = 3 s. Zasyp liczony jako blok z odsypami w trakcie (mediana 3 s na granicach zasypu).
- Grupy: **kontrolna** (bez zdarzeń: R = sama różnica dryfu noc do nocy), **bez zasypu**, **z zasypem**. Doby z zakłóceniem oznaczone, nie wyrzucane.

Czas lokalny (Europe/Warsaw); przeglądarka dób podaje też UTC. Plan: `plans/dms-data-analysis/etap3-miary-bledu-plan.md` §4. Obliczenia: `uv run python -m dms_analysis.errors balance`.
