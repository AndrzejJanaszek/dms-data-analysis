# Etap 3, część 3 — prawdziwy harmonogram: suma dnia i odtworzone odsypy

Bierzemy prawdziwy dzień pracy (**harmonogram**: godzina startu i czas trwania każdego odsypu) i przenosimy wszystkie odsypy na te same godziny zegarowe **doby bazowej** — pełnej doby bez żadnych zdarzeń. Na wskazaniach doby bazowej każdy odsyp ma masę 0, więc wynik metody (przed − po) to sam błąd z dryfu i szumu. **E** = suma błędów odsypów dnia, **Y** = suma prawdziwych mas odsypów harmonogramu (poza zasypem), **E / Y** = błąd sumy dnia względem tego, co wydano.

Para (harmonogram, doba bazowa) ważna, gdy wszystkie odsypy z oknami leżą we fragmentach bez zdarzeń (60 s od nich), problemów z katalogu i luk. Doby tylko z okresów z czasem rzeczywistym; 2.11 sklejona z dwóch okresów (przerwa 116 s, zmiana wskazania −20 kg). Plan: `plans/dms-data-analysis/etap3-miary-bledu-plan.md` §5. Obliczenia: `uv run python -m dms_analysis.errors schedule`.

## 5. Porównanie z bilansem doby

Dla tych samych dób: R z bilansu noc–noc (część 2) i E — ile dałby sam dryf spokojnych dób przy tym samym harmonogramie. **Znak jest ten sam:** R = dryf całej doby + Σ błędów odsypów − Σ błędów zasypów, a E = Σ błędów odsypów na dobie bazowej, więc w obu dodatni wynik = masy odsypów wyszły za duże, ujemny = za małe. Doba 2.11 nie ma R, bo bilans wymaga jednej części okresu (2.11 jest sklejona z dwóch okresów).

```python
views.balance_comparison(tables, balances)
```
