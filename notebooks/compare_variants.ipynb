# Porównanie wariantów przygotowania sygnałów

Te same dane liczone dwoma wariantami reguł (`VARIANTS` w `dms_analysis.signals`). Zdarzenia A i B tego samego pinu nakładające się w czasie tworzą grupę; każda grupa dostaje kategorię:

- **bez zmian** — ten sam początek, koniec i status,
- **zmiana statusu** — te same granice, ale w jednym wariancie przyjęte, w drugim odrzucone,
- **zmienione granice** — jedno zdarzenie w A i jedno w B, inny początek albo koniec,
- **połączone w B** / **rozdzielone w B** — kilka zdarzeń A to jedno w B albo odwrotnie,
- **przegrupowane** — kilka po obu stronach.

W przeglądarce pod wskazaniami: sygnał surowy, wariant A i wariant B; tło w wierszu wariantu: zielone = przyjęte, czerwone = odrzucone.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from IPython.display import display

from dms_analysis.browser import DiffBrowser
from dms_analysis.compare import summary
from dms_analysis.raw import load_session
from dms_analysis.signals import PIN_ROLES, VARIANTS, detect_events

DB_PATH = Path("../data/meas-signs-temps_2026-10-18_no-temps-problem/measurement-data-saver.db")
SESSION_ID = 9
VARIANT_A = "v3"
VARIANT_B = "v4"
FIGURE_WIDTH = 1200
```

```python
data = load_session(DB_PATH, SESSION_ID)
events_a = detect_events(data.pins, data.end, variant=VARIANTS[VARIANT_A])
events_b = detect_events(data.pins, data.end, variant=VARIANTS[VARIANT_B])

for name in (VARIANT_A, VARIANT_B):
    print(f"{name}: {VARIANTS[name].description}")
    for pin, role in PIN_ROLES.items():
        print(f"  pin {pin} ({role}): {VARIANTS[name].rules(pin)}")
```

## Różnice

```python
diff = DiffBrowser(
    data, events_a, events_b, (VARIANT_A, VARIANT_B), f"{DB_PATH.parent.name}, sesja {SESSION_ID}", FIGURE_WIDTH
)
summary(diff.clusters)
```
