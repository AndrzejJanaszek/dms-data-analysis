# Przeglądarka zdarzeń

Przegląd faktycznych danych z wybranych przedziałów histogramów z Grafany („Histogram czasu trwania zdarzeń” i „Histogram przerw między kolejnymi zdarzeniami na pinie”), przed zmianą parametrów przygotowania sygnałów.

Widok: wskazania oraz sygnał pinu zdarzenia, surowy i po obróbce, w oknie zdarzenia z marginesem z obu stron. Tło: zielone = zdarzenie przyjęte, czerwone = odrzucone krótkie, pomarańczowe = przerwa. Przedział `od`–`do` jest lewostronnie domknięty, jak kubełki w Grafanie. Parametry pipeline'u to wariant z `VARIANTS` w `dms_analysis.signals` (domyślnie `CURRENT_VARIANT`); inny wariant można wybrać w komórce niżej i przeliczyć zdarzenia.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from IPython.display import display

from dms_analysis.browser import EventBrowser, events_by_duration, events_by_gap
from dms_analysis.raw import load_session
from dms_analysis.signals import CURRENT_VARIANT, PIN_ROLES, VARIANTS, detect_events, summary

DB_PATH = Path("../data/meas-signs-temps_2026-10-18_no-temps-problem/measurement-data-saver.db")
SESSION_ID = 9
VARIANT = CURRENT_VARIANT
FIGURE_WIDTH = 1200
```

```python
data = load_session(DB_PATH, SESSION_ID)
variant = VARIANTS[VARIANT]
signal_events = detect_events(data.pins, data.end, variant=variant)
title = f"{DB_PATH.parent.name}, sesja {SESSION_ID}, wariant {VARIANT}"

for pin, role in PIN_ROLES.items():
    print(f"pin {pin} ({role}): {variant.rules(pin)}")
summary(signal_events)
```

## Zdarzenia o czasie trwania w przedziale

Przyjęte i odrzucone razem, zakres okna: zdarzenie ± margines.

```python
durations = EventBrowser(data, signal_events, events_by_duration, 1, 2, title, FIGURE_WIDTH)
display(durations.widget)
```

## Pary zdarzeń z przerwą w przedziale

Przerwa = początek zdarzenia minus koniec poprzedniego zdarzenia tego samego pinu (przyjęte i odrzucone razem). Zakres okna: od początku pierwszego do końca drugiego zdarzenia ± margines.

```python
gaps = EventBrowser(data, signal_events, events_by_gap, 1, 3, title, FIGURE_WIDTH)
display(gaps.widget)
```
