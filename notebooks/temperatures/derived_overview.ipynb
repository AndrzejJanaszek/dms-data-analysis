# Dane pochodne temperatur — zestawienia

Wyniki `uv run python -m dms_analysis.temperatures compute` (tabela `temperatures`, opis kolumn w README). Ustaw, co jest na górnym panelu (`TOP`: wskazanie wagi albo temperatury czujników), okres, przedział (czas lokalny) i krok próbkowania: dla całej doby 5 s, dla krótkich fragmentów 1 s.

```
from dms_analysis import style
from dms_analysis.data.catalog import load_catalog
from dms_analysis.temperatures.views import all_figures, load_range, load_weight

style.register()
```

```
TOP = "wskazanie"
PERIOD = "swietochlowice-1/0005-2026-11-02"
START = "2026-11-07 00:00"
END = "2026-11-08 00:00"
STEP_S = 5

period = load_catalog().period(PERIOD)
corners = {sensor.number: sensor.corner for sensor in period.hardware.temperature_sensors if sensor.corner}
frame = load_range(PERIOD, START, END, STEP_S)
weight = load_weight(PERIOD, START, END, STEP_S) if TOP == "wskazanie" else None
figures = all_figures(frame, corners, f"{PERIOD}, {START} – {END} lok., krok {STEP_S} s", weight)
list(figures)
```
